```python
import numpy as np
import pandas as pd
df=pd.read_csv("WA_Fn-UseC_-HR-Employee-Attrition.csv")
print(df.isnull().sum())
print("Duplicates:",df.duplicated(subset=['EmployeeNumber']).sum())
```

```python
print("Shape:",df.shape)
print("Duplicate rows:",df.duplicated().sum)
```

```python
df['Attrition_Flag']=np.where(df['Attrition']=='Yes',1,0)
```

```python
df['TenureBand']=pd.cut(df['YearsAtCompany'],bins=[-1,2,5,10,40],labels=['0-2 Yrs','3-5 Yrs','6-10 Yrs','10+ Yrs'])
```

```python
df['PayBand']=pd.qcut(df['MonthlyIncome'],q=4, labels=['Low','Lower-Mid','Upper-Mid','High'])
```

```python
print(df.groupby('JobRole')[['Attrition_Flag','MonthlyIncome']].mean())
```

```python
print(np.percentile(df['MonthlyIncome'],[25,50,75,90]))
```

```python
print(np.mean(df['MonthlyIncome']) - 1.96 * (np.std(df['MonthlyIncome'],ddof=1) / np.sqrt(len(df))), np.mean(df['MonthlyIncome']) + 1.96 * (np.std(df['MonthlyIncome'], ddof=1) /np.sqrt(len(df))))
```

```python
print(df.groupby(['Department','TenureBand','OverTime'],observed=False)['Attrition_Flag'].mean())
```

```python
print(df.groupby('Department')['MonthlyIncome'].agg(['mean','std']))
```

```python
q75, q25 = np.percentile(df['MonthlyIncome'],[75, 25])
```

```python
print(df.groupby('PayBand',observed=False)['Attrition_Flag'].mean())
```

```python
print(df.groupby(['PayBand','OverTime'], observed=False)['Attrition_Flag'].mean())
```
